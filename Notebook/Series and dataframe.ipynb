### Pandas Series

A Series is a one-dimensional labeled array.

Think of it like:

A single column of a table

Or a vector with labels (index)

```python
import pandas as pd
```

```python
s = pd.Series([10, 20, 30, 40])
s
```

### Giving a Series your own index

By default a Series is labelled `0, 1, 2, ...`. Passing `index=[...]` replaces
those numbers with labels you choose, which is what makes `marks["Sita"]` possible.

```python
#series with custom index
marks = pd.Series([85, 90, 88], index=["Ananda", "Sita", "Ram"])
marks
```

```python
# Access by POSITION -> use .iloc
marks.iloc[0]
```

```python
import pandas as pd
import random

# Seed the generator so this notebook produces the SAME file every time it runs
random.seed(42)

names = [
    "Ram", "Shyam", "Hari", "Sita", "Gita", "Ramesh", "Suresh",
    "Bikash", "Prakash", "Anil", "Sunil", "Aashish", "Bibek",
    "Nabin", "Suman", "Kiran", "Sabin", "Roshan", "Manish", "Dipesh"
]

fields = ["IT", "Medicine", "Engineer", "Business", "Finance", "Education"]

cities = [
    "Kathmandu", "Pokhara", "Lalitpur", "Bhaktapur", "Chitwan",
    "Biratnagar", "Dharan", "Butwal", "Janakpur", "Nepalgunj"
]

data = {
    "Name": [random.choice(names) for _ in range(1000)],
    "Salary": [random.randint(20000, 150000) for _ in range(1000)],
    "Field": [random.choice(fields) for _ in range(1000)],
    "Age": [random.randint(20, 60) for _ in range(1000)],
    "City": [random.choice(cities) for _ in range(1000)]
}

df = pd.DataFrame(data)

df
```

### Reading the file back

A good habit after saving: read the file again and confirm it looks the way you expect.

```python
# Read the file we just wrote
check = pd.read_csv("../data/nepali_data.csv")
print(check.shape)
check.head()
```

## Task — detailed comparison between Series and DataFrame

> *Original note: "find and write the detailed comparison between Series and DataFrame with code example"*

| | **Series** | **DataFrame** |
|---|---|---|
| Dimensions | 1-D | 2-D |
| Think of it as | one column | a whole table |
| Axes | one index | an index **and** columns |
| `.shape` | `(n,)` | `(rows, cols)` |
| dtype | one dtype for the whole Series | one dtype **per column** |
| Built from | a list, a dict, `df["col"]` | a dict of lists, a list of dicts, `df[["col"]]` |
| Has `.columns`? | no | yes |
| `.name` | the column's name | not applicable |
| A single column of it | `s.iloc[0]` → one value | `df["col"]` → a Series |
| Common use | one variable, one measurement | the dataset itself |

**The link between them:** a DataFrame is a collection of Series that share one index.
Pull a column out and you get a Series; put Series back together and you get a
DataFrame.

```python
import pandas as pd

# The SAME data, once as a Series and once as a DataFrame
scores_series = pd.Series([85, 90, 88], index=["Ananda", "Sita", "Ram"], name="Score")
scores_frame  = pd.DataFrame({"Score": [85, 90, 88]}, index=["Ananda", "Sita", "Ram"])

print("SERIES"); print(scores_series)
print()
print("DATAFRAME"); print(scores_frame)
```

```python
# 1. Type and shape
print("type :", type(scores_series).__name__, "|", type(scores_frame).__name__)
print("shape:", scores_series.shape, "     |", scores_frame.shape)
print("ndim :", scores_series.ndim, "              |", scores_frame.ndim)
```

```python
# 2. A Series has a .name, a DataFrame has .columns
print("Series .name    :", scores_series.name)
print("DataFrame .columns:", list(scores_frame.columns))
```

```python
# 3. Selecting one element
print("Series by label   :", scores_series["Sita"])
print("Series by position:", scores_series.iloc[1])
print("DataFrame needs BOTH row and column:", scores_frame.loc["Sita", "Score"])
```

```python
# 4. dtypes: one for a Series, one PER COLUMN for a DataFrame
mixed = pd.DataFrame({
    "Name":  ["Ananda", "Sita", "Ram"],
    "Age":   [22, 21, 23],
    "Score": [85.5, 90.0, 88.5]
})
print("Series dtype :", scores_series.dtype)
print()
print("DataFrame dtypes:")
print(mixed.dtypes)
```

```python
# 5. Going back and forth
col = mixed["Age"]                 # DataFrame -> Series
back = col.to_frame()              # Series -> DataFrame
rebuilt = pd.DataFrame({"Age": mixed["Age"], "Score": mixed["Score"]})  # Series -> DataFrame

print(type(col).__name__, "->", type(back).__name__)
rebuilt
```

```python
# 6. The same method behaves differently on each
print("Series.describe() — one summary:")
print(scores_series.describe())
print()
print("DataFrame.describe() — one summary PER numeric column:")
print(mixed.describe())
```

### Key takeaways

- `df["col"]` → **Series** · `df[["col"]]` → **DataFrame** (one column, still a table)
- A DataFrame is just Series glued together on a shared index
- A Series has one dtype; a DataFrame has one per column
- Selecting from a Series needs **one** key; from a DataFrame, a row **and** a column
