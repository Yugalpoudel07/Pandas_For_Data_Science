# Indexes in Pandas — Set, Reset and Use

The **index** is the label of each row. By default it is `0, 1, 2, ...`, but any
column can become the index.

You will learn:

- `df.index` — what the current labels are
- `set_index()` — make a column the row label
- `reset_index()` — turn the index back into a normal column
- `inplace=True` vs reassigning `df = ...`
- Why a meaningful index makes `.loc` lookups readable

````python
# 1️⃣ Import Pandas Library
# Import pandas to work with DataFrames
import pandas as pd
````

````python
# 2️⃣ Create a Sample DataFrame
# We will create a DataFrame with columns 'Name', 'Age', 'City'
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['Kathmandu', 'Pokhara', 'Lalitpur', 'Biratnagar']
}

df = pd.DataFrame(data)  # Create DataFrame
df
````

````python
# 3️⃣ Default Index
# By default, Pandas assigns a numeric index starting from 0
print(df.index)  # Shows the default index
````

# 4️⃣ Set a Column as Index

````python
df.set_index("Name")
````

### 5. Making the change permanent

`set_index()` returns a **new** DataFrame and leaves the original alone.
To keep the change you either reassign (`df = df.set_index("Name")`) or pass
`inplace=True`.

````python
df.set_index("Name",inplace=True)
````

````python
df
````

````python
# 6️⃣ Reset Index

df.reset_index()

````

````python
# 6️⃣ Reset Index

df.reset_index(inplace=True)

````

### What does inplace=True mean?

- `inplace=True` means **modify the original DataFrame**
- No new DataFrame is created
- Changes happen **permanently**

## Task — create your own data and practise the whole cycle

Below: build a small DataFrame, set an index without `inplace`, set it **with**
`inplace`, look rows up by label, then reset it back.

````python
import pandas as pd

# A small table of shops
shops = pd.DataFrame({
    "ShopID": ["S01", "S02", "S03", "S04"],
    "Owner":  ["Anita", "Bibek", "Chandra", "Dipesh"],
    "City":   ["Kathmandu", "Pokhara", "Butwal", "Dharan"],
    "Sales":  [120000, 98000, 143000, 76000]
})
shops
````

````python
# Step 1 — the default index is RangeIndex 0,1,2,3
shops.index
````

````python
# Step 2 — set_index WITHOUT inplace: returns a new DataFrame
shops.set_index("ShopID")
````

````python
# ...and the original is untouched, the index is still 0,1,2,3
shops.index
````

````python
# Step 3 — now do it permanently with inplace=True
shops.set_index("ShopID", inplace=True)
shops
````

````python
# Step 4 — the labels are now the shop ids, so .loc reads like plain English
shops.loc["S03"]
````

````python
# One value: .loc[row_label, column_label]
shops.loc["S03", "Owner"]
````

````python
# Several labels at once
shops.loc[["S01", "S04"], ["Owner", "Sales"]]
````

````python
# .iloc still works — it ignores the labels and counts POSITIONS
print("Position 0 label:", shops.iloc[0].name)
shops.iloc[0]
````

````python
# Step 5 — reset_index() puts ShopID back as a normal column
shops.reset_index(inplace=True)
shops
````

````python
# reset_index(drop=True) throws the old index away instead of keeping it.
# Handy after filtering, when the leftover index has gaps like 3, 7, 12.
filtered = shops.loc[shops["Sales"] > 90000]
print("Index after filtering:", list(filtered.index))
filtered = filtered.reset_index(drop=True)
print("Index after reset_index(drop=True):", list(filtered.index))
filtered
````

### Key takeaways

| Goal | Code |
|------|------|
| See the current labels | `df.index` |
| Make a column the index | `df.set_index("col")` |
| Keep that change | `inplace=True` or `df = df.set_index("col")` |
| Look a row up by label | `df.loc["S03"]` |
| Look a row up by position | `df.iloc[0]` |
| Index back to a column | `df.reset_index()` |
| Renumber and discard the old index | `df.reset_index(drop=True)` |
