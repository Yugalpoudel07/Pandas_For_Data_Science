```python
import pandas as pd

data = {
    "Name": ["Ram", "Sita", "Hari", "Gita"],
    "Age": [22, 19, 25, 21],
    "Marks": [78, 85, 67, 90]
}

df = pd.DataFrame(data)
df
```

```python
df.head(3)
```

```python
df.tail(3)
```

```python
#Sort by Age
df.sort_values(by="Age")
```

```python
# Sort by Marks
df.sort_values(by="Marks")
```

```python
# First by Age, then by Marks
df.sort_values(by=["Age", "Marks"])
```

```python
df.sort_values(by=["Age", "Marks"], ascending=[True, False])
# Different order for each column:
```

```python
df.sort_index()
# sort by index
```

```python
df.sort_index(ascending=False)
```

```python
df
```

```python
# Sort and save changes (inplace)
df.sort_values(by="Marks", inplace=True)
df
```

```python
df
```

```python
# Sort strings (alphabetical order)
df.sort_values(by="Name")
```

```python
# Reverse alphabetical:
df.sort_values(by="Name", ascending=False)
```

## 🧠 Quick Summary

| Task | Method |
|------|--------|
| Sort rows | `sort_values()` |
| Sort index | `sort_index()` |
| Descending order | `ascending=False` |
| Multiple columns | `by=[...]` |
| Save changes | `inplace=True` |
| Fix index | `reset_index(drop=True)` |

## Practice — sorting on a real dataset

The examples above used a 4-row table. Here are the same methods on the
1000-row `nepali_data.csv`, where sorting actually tells you something.

```python
import pandas as pd

emp = pd.read_csv("../data/nepali_data.csv")
emp.head()
```

```python
# Highest paid first
emp.sort_values(by="Salary", ascending=False).head(10)
```

```python
# Lowest paid first (ascending is the default)
emp.sort_values(by="Salary").head(10)
```

```python
# Two columns: group the fields alphabetically, and inside each field
# put the highest salary on top.
emp.sort_values(by=["Field", "Salary"], ascending=[True, False]).head(15)
```

```python
# Sorting does NOT renumber the rows — the old index travels with them.
# reset_index(drop=True) gives a clean 0,1,2,... again.
top = emp.sort_values("Salary", ascending=False).head(5)
print("Index straight after sorting:", list(top.index))
top = top.reset_index(drop=True)
print("After reset_index(drop=True) :", list(top.index))
top
```

```python
# Sorting a column that has missing values:
# by default NaN goes LAST, na_position="first" flips it.
with_gaps = emp.copy()
with_gaps.loc[0:4, "Salary"] = None

print("NaN last (default):")
print(with_gaps.sort_values("Salary")[["Name", "Salary"]].tail(3))
print()
print('NaN first (na_position="first"):')
print(with_gaps.sort_values("Salary", na_position="first")[["Name", "Salary"]].head(3))
```

```python
# sort_index() sorts by the ROW LABELS rather than by a column.
# Most useful right after a filter, to put the rows back in their original order.
sample = emp.sample(8, random_state=1)
print("Random order:", list(sample.index))
sample.sort_index()
```

```python
# Sorting a groupby result is the usual way to build a "top N" table
emp.groupby("Field")["Salary"].mean().sort_values(ascending=False)
```
