# Pandas Demo — First Look at a Dataset

The very first thing you do with any new dataset: load it and look at it.

You will learn:

- `read_csv()` — load a CSV into a DataFrame
- `head()` / `tail()` — peek at the first and last rows
- `shape` — how many rows and columns
- `info()` — dtypes and missing-value counts
- `describe()` — quick statistics

Dataset: `../data/titanic_dataset.csv`

```python
import pandas as pd
```

```python
df = pd.read_csv('../data/titanic_dataset.csv')
```

```python
df.head(15)
```

```python
import pandas as pd

# A completely different dataset: 1000 generated people
emp = pd.read_csv("../data/nepali_data.csv")
emp.head()
```

```python
# First 10 and last 10 rows
emp.head(10)
```

```python
emp.tail(10)
```

```python
# (rows, columns)
emp.shape
```

```python
# dtypes + non-null counts. Every column here is full, so there is nothing missing.
emp.info()
```

```python
# Statistics for the numeric columns (Salary, Age)
emp.describe()
```

```python
# include="object" switches describe() to the text columns.
# "top" is the most frequent value, "freq" is how often it appears.
emp.describe(include="object")
```

```python
# A few more first-look checks
print("Columns      :", list(emp.columns))
print("Missing cells:", emp.isna().sum().sum())
print("Duplicates   :", emp.duplicated().sum())
print("Fields       :", emp["Field"].nunique())
emp["City"].value_counts()
```

### Key takeaways

| Question | Code |
|----------|------|
| What does the data look like? | `df.head()` / `df.tail()` |
| How big is it? | `df.shape` |
| What types, what is missing? | `df.info()` |
| Numeric summary | `df.describe()` |
| Text summary | `df.describe(include="object")` |
| Any missing / duplicated? | `df.isna().sum()` / `df.duplicated().sum()` |
