# Pandas `loc` vs `iloc`
## Detailed Comparison with Examples


````python
import pandas as pd
````

````python
data = {
    "ID": [101, 102, 103, 104],
    "Name": ["Ram", "Sita", "Hari", "Gita"],
    "Marks": [85, 90, 78, 92]
}

df = pd.DataFrame(data)
df
````

## `loc` – Label-based Indexing

- `loc` is used to access data using **labels**
- Labels can be:
  - Index names /row label
  - Column names
- **Both start and end values are included**


````python
df.set_index("ID", inplace=True)
df
````

loc=access data using label index name value not index postion

````python
df.loc[101,"Name"]
````

````python
df.loc[[101, 103]] # access multiple rows
````

````python
df.loc[101] #Access Specific rows
````

````python
df
````

````python
df.loc[101:103, ["Name", "Marks"]]
````

## `iloc` – Integer Position-based Indexing

- `iloc` uses **integer positions**
- Index starts from **0**
- **End value is excluded**
- Similar to Python list slicing


````python
df.set_index("Marks")
````

````python
df
````

````python
df
````

````python
df.iloc[[0, 2]] #multiple rows
````

````python
df.iloc[1, 0] #Access Specific Row & Column
````

````python
df.iloc[0:3, 0:2] #Access Range of Rows & Columns
````

## Comparison Table

| Feature | loc | iloc |
|------|-----|------|
| Type | Label-based | Integer position-based |
| Uses | Index & column names | Row & column numbers |
| Range | End included | End excluded |
| Similar to | Dictionary | Python list |
| Example | df.loc[101] | df.iloc[0] |

## 📌 10. When to Use What? (Markdown Cell)

```markdown
### When to Use `loc`
- When index has meaningful labels (ID, Name, Date)
- When readability is important

### When to Use `iloc`
- When working with positions
- When index values are unknown
- For fast slicing like arrays


````python
# Quick self-check: predict the output of each line before running it.
import pandas as pd

students = pd.DataFrame(
    {"Name": ["Anita", "Bibek", "Chandra", "Dipesh"],
     "Marks": [85, 72, 91, 64]},
    index=[101, 102, 103, 104]        # the index is NOT 0,1,2,3
)
print(students)
print()
print("loc[101]  -> label 101 :", students.loc[101, "Name"])
print("iloc[0]   -> position 0:", students.iloc[0]["Name"])
print()
print("loc[101:103]  ->", len(students.loc[101:103]), "rows (end label INCLUDED)")
print("iloc[0:3]     ->", len(students.iloc[0:3]), "rows (end position EXCLUDED)")
````

### Key takeaways

- `.loc` = **labels**, end **included**
- `.iloc` = **positions**, end **excluded**
- They only look the same while the index happens to be `0, 1, 2, ...`
- Use `.loc` whenever you are **writing** values: `df.loc[cond, "col"] = value`
